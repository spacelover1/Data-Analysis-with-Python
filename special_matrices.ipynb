```python
import numpy as np
```

# Row Matrix

```python
row = np.array([[1, 2, 3, 4, 5]])
row
```

# Column Matrix

```python
col = row.reshape(-1, 1)
col
```

```python
col.shape
```

# Squared Matrix

```python
sq = np.array([[1, 2, 5], [3, 4, 6], [7, 8, 9]])
sq
```

# Diagonal Matrix

```python
diag = np.diag([1, 2, 3, 4])
diag
```

```python
np.diag(sq)
```

```python
np.diag(np.fliplr(sq))
```

# Ones

```python
ones = np.ones((3, 3), dtype=np.int16)
ones
```

## Ones-like

```python
np.ones_like(sq)
```

# Zeros

```python
zeros = np.zeros((4, 4))
zeros
```

## Zeros-like

```python
np.zeros_like(sq)
```

# Full Matrix

```python
full = np.full((4, 5), "Honey")
full
```

## Full-like

```python
np.full_like(sq, 12)
```

# Upper Triangular

```python
np.triu(sq)
```

```python
np.triu(sq, k=1)
```

```python
np.triu(sq, k=-1)
```

# Lower Triangular

```python
np.tril(sq)
```

# Identity
