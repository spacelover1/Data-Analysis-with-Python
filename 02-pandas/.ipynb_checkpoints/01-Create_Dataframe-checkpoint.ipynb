```python
info = [
    ["DODGE RAM", 800, 80_000, "Red", 2017],
    ["Nissan GTR", 1400, 90_000, "Pink", 2019],
    ["Dena", 170, 10_000, "White", 2022],
    ["206", 72, 15_000, "Black", 2025],
    ["BMW M4", 500, 40_000, "Grey", 2026]
]

df3 = pd.DataFrame(data=info, columns=['Model', 'HorsePower', 'Price', 'Color', 'Year'], index=range(1,6))
df3
```
