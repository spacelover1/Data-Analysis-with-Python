### Third: performance and runtime

```python
import sys
import time
from sys import getsizeof

start = time.time()

x = list(range(100000000))
for i in range(100000000):
    x[i] += 1


end = time.time()
print(end-start, "seconds")
print(sys.getsizeof(x)/1024*1024, "MB")
```

```python
import time
import sys

start = time.time()

x = np.arange(100000000) + 1



end = time.time()
print(end - start, "seconds")
print(sys.getsizeof(x))
```
