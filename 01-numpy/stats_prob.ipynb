# Measures of Central Tendency 
## Mean

mean(avg) = sum of values/ number of values >>>   x̅ =  (x1 + x2 + x3 + . . . + xn) / n

### What is the difference between MU(μ) and X bar (x̅)?

Mu (μ) and X-bar (x̄) are both measures of central tendency—the mean—but they refer to different contexts and have distinct statistical roles.\
μ (mu): population mean. The arithmetic average of every individual value in the entire population. It is a fixed but usually unknown parameter.\
x̄ (X-bar): sample mean. The arithmetic average of values in a sample drawn from the population. It is a statistic and therefore a random variable that varies from sample to sample.

## Median

A median is a middle value for sorted data.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
heights = [169, 175, 170, 175, 160, 174, 170, 170, 160, 157, 172, 182, 208, 130]
len_heights = range(len(heights))
len_heights
```

```python
plt.scatter(len_heights, heights);
```

```python
mean_heights = np.mean(heights)
plt.scatter(len_heights, heights, label= "Heights", c="Black");
plt. plot([0, 14], [mean_heights, mean_heights], label="Mean", color="Red");
plt.legend()
```

# Measures of dispersion 

Quantify how spread out or varied values are in a dataset.

## Standard Deviation
## Variance

```python
std = np.std(heights)
plt.scatter(len_heights, heights, label= "Heights", c="Black");
plt. plot([0, 14], [mean_heights, mean_heights], label="Mean", color="Red");
plt. plot([0, 14], [mean_heights+std, mean_heights+std], label="Mean+std", color="Orange");
plt. plot([0, 14], [mean_heights-std, mean_heights-std], label="Mean-std", color="Green");
plt.legend()
```

# Common Statistical Distributions

## Binomial Distribution
Probability of a binary outcome in an experiment that we run multiple times

## Bernoulli Distribution 
(special case of binomial dist) here, instead of considering all possible outcomes, we're just considering two possible outcome>> (success/failure | true/fals | yes/no)

## Uniform Distribution 
All outcomes are equally likely to occur 

## Poisson Distribution
(bounded between zero and infinity so not symmetrical) Describes the number of outcomes that occur during some fixed interval, most commonly a time interval.
